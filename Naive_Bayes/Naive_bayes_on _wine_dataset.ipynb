```python
#import libraries

import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
# Load the wine dataset
wine = load_wine()
X, y = wine.data, wine.target
```

```python

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
# Train a Gaussian Naive Bayes classifier
gnb = GaussianNB()
gnb.fit(X_train, y_train)
y_pred_gnb = gnb.predict(X_test)
accuracy_gnb = accuracy_score(y_test, y_pred_gnb)
print(f"Gaussian Naive Bayes accuracy: {accuracy_gnb}")
```

```python
# Train a Multinomial Naive Bayes classifier
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
y_pred_mnb = mnb.predict(X_test)
accuracy_mnb = accuracy_score(y_test, y_pred_mnb)
print(f"Multinomial Naive Bayes accuracy: {accuracy_mnb}")
```

```python
# Compare the performance of the two models
if accuracy_gnb > accuracy_mnb:
    print("Gaussian Naive Bayes performs better.")
    best_model = gnb
    y_pred = y_pred_gnb
else:
    print("Multinomial Naive Bayes performs better.")
    best_model = mnb
    y_pred = y_pred_mnb
```
